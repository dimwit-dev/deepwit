```scala
{{
val (steps, losses) = logger.loadTensor0("train/loss").get

val trace = Scatter()
  .withX(toPyTensor(steps).as[Seq[Int]])
  .withY(toPyTensor(losses).as[Seq[Double]])
  .withMode(ScatterMode(ScatterMode.Lines))
  .withName("Training Loss")

val layout = Layout()
  .withTitle("Interactive Training Loss")
  .withXaxis(plotly.layout.Axis().withTitle("Iteration"))
  .withYaxis(plotly.layout.Axis().withTitle("Loss"))

plot(Seq(trace), layout)
}}
```
